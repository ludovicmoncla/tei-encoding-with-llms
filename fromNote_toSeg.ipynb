## Notebook permettant de remplacer l'éléménet `<note>` par `<seg>`, un changement réalisé au cours de nos expérimentations.

```python
import os
from lxml import etree
from pathlib import Path
import re
import glob
```

```python
def from_cit_to_gloss(from_path, from_pattern, to_pattern, to_path):
    for file in from_path:
        with open(file, "r", encoding="utf-8") as f:
            content=f.read()
        new_content=from_pattern.sub(to_pattern, content)

        chemin_sortie=to_path/file.name
        with open(chemin_sortie, "w", encoding="utf-8") as f:
            f.write(new_content)
```

### From `<note>` to `<seg>`

```python
note_to_seg_from_path=(sorted(Path("gt/niveau3").glob("*.xml")))
note_to_seg_to_path=Path("gt/niveau3_new")
note_to_seg_to_path.mkdir(parents=True, exist_ok=True)
note_pattern = re.compile(r'<note>(.*?)</note>', re.DOTALL)
seg_pattern=r'<seg>\1</seg>'
from_cit_to_gloss(note_to_seg_from_path, note_pattern, seg_pattern, note_to_seg_to_path)
```
